string concatenation

```python
string1='hello'
string2=input('Enter your name: ')
string=string1+' '+string2
print(string)
```

```python
string3=', welcome to python programming'
string4=string+string3
print(string4)
```

string slicing and indexing

```python
print(string4[0])
print(string4[-1])
print(string4[0:5])
print(string4[-11:])
print(string4[::-1])
print(string4[-18:-12])
```

string methods

```python
strM="Python beginner tutorial"
print(strM.upper())
print(strM.lower())
print(strM.capitalize())
print(strM.count("t"))
print(strM.replace('Python','Machine Learning'))
```

```python
tuple1=(10,20,30)
tuple2=(40,50,60)
t_combine=tuple1+tuple2
print(t_combine)
print(t_combine*3)
print(t_combine[2])
print(t_combine[0:3])
print(t_combine[-3:])
```
